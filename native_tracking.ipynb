```python
import torch
from otx.backend.native.models.detection.yolox import YOLOX
from otx.backend.native.models.base import DataInputParams
from otx.backend.native.models.tracking.bytetrack import ByteTrack

DEVICE = "mps" if torch.backends.mps.is_available() else "cpu"

# Model = detector (any OTXDetectionModel)
model = YOLOX(
    label_info=80,
    data_input_params=DataInputParams(
        input_size=(416, 416),
        mean=(123.675, 116.28, 103.53),
        std=(58.395, 57.12, 57.375),
    ),
    model_name="yolox_tiny",
)
model.eval()
model.to(DEVICE)

# Tracker = ByteTrack (detector-agnostic)
tracker = ByteTrack(track_thresh=0.5, track_buffer=30, match_thresh=0.8)

print(f"Model: {model.model_name} | Device: {DEVICE}")
```

```python
from IPython.display import Video, display

# Track a video — model passed at call time
out = tracker.track(model, "videos/original/jets-2.mp4", output_dir="videos/otx-tracked", device=DEVICE)
display(Video(str(out), embed=True, width=800))
```
